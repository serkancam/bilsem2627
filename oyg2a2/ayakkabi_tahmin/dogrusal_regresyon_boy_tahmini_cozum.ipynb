# ÇÖZÜM: Doğrusal Regresyon ile Boy Tahmini

Bu dosya, öğrenci görevindeki eksik kısımların tamamlanmış **çözüm (cevap anahtarı)** versiyonudur.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_theme(style="whitegrid")
```

```python
# Veriyi yükleme ve ön inceleme
df = pd.read_excel('ayakkabi_numarasi.xlsx')

# Cinsiyet sütununu sayısal (1 ve 0) formata çevirme
df['Cinsiyet_E'] = (df['Cinsiyet(E veya K)'] == 'E').astype(int)
df.head()
```

### ÇÖZÜM 1: X ve y Değişkenlerinin Ayrılması

```python
# X için 'Kilo(kg)', 'Yaş', 'Cinsiyet_E', 'Ayakkabı Numarası' sütunlarını seçin
feature_cols = ['Kilo(kg)', 'Yaş', 'Cinsiyet_E', 'Ayakkabı Numarası']
X = df[feature_cols]

# y için 'Boy(metre)' sütununu seçin
y = df['Boy(metre)']

print("X boyutu:", X.shape)
print("y boyutu:", y.shape)
```

### ÇÖZÜM 2: Veriyi Eğitim ve Test Setlerine Ayırma

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20,
    random_state=42
)
```

### ÇÖZÜM 3: Modelin Kurulması ve Eğitilmesi

```python
# Modeli çağırın
model = LinearRegression()

# Modeli X_train ve y_train verileri ile eğitin
model.fit(X_train, y_train)

print("Model başarıyla eğitildi!")
```

### ÇÖZÜM 4: Tahmin Yapma ve Metriklerin Hesaplanması

```python
# Test seti üzerinde tahmin yapın
y_pred = model.predict(X_test)

# R² Skorunu hesaplayın
r2 = r2_score(y_test, y_pred)
print(f"Test Seti R² Skoru: {r2:.3f}")
```
