# ÇÖZÜM: Polinomsal Regresyon ile Boy Tahmini

Bu dosya, öğrenci görevindeki eksik kısımların tamamlanmış **çözüm (cevap anahtarı)** versiyonudur.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

df = pd.read_excel('ayakkabi_numarasi.xlsx')
df['Cinsiyet_E'] = (df['Cinsiyet(E veya K)'] == 'E').astype(int)
```

### ÇÖZÜM 1: Değişkenlerin Ayrılması ve Split İşlemi

```python
df.columns
```

```python
# Bağımsız değişkenleri (X) belirtin
X = df[['Kilo(kg)', 'Yaş', 'Cinsiyet_E', 'Ayakkabı Numarası']]

# Hedef değişkeni (y) belirtin
y = df['Boy(metre)']

# Veriyi %80 Eğitim, %20 Test olarak ayırın
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

### ÇÖZÜM 2: Polinomsal Dönüşüm (Polynomial Features)

```python
# PolynomialFeatures sınıfından 2. derece nesne oluşturun
poly_converter = PolynomialFeatures(degree=2, include_bias=False)

# Eğitim verisini dönüştürün
X_train_poly = poly_converter.fit_transform(X_train)

# Test verisini dönüştürün
X_test_poly = poly_converter.transform(X_test)
```

### ÇÖZÜM 3: Modelin Kurulması, Eğitilmesi ve Test Edilmesi

```python
model = LinearRegression()

# Modeli dönüştürülmüş EĞİTİM verisi ile eğitin
model.fit(X_train_poly, y_train)

# Modeli dönüştürülmüş TEST verisi ile tahmin ettirin
y_pred = model.predict(X_test_poly)

print("Test Seti R² Skoru:", r2_score(y_test, y_pred))
```
