# ÖĞRENCİ GÖREVİ: Doğrusal Regresyon ile Boy Tahmini

Sevgili öğrenciler, bu çalışmada hedef değişkenimiz **Boy(metre)** olarak değiştirilmiştir. Amacımız; Kilo, Yaş, Cinsiyet ve Ayakkabı Numarası verilerini kullanarak bir kişinin boyunu tahmin etmektir.

Aşağıdaki kod hücrelerinde bazı kısımlar **eksik bırakılmıştır (`___` veya `# EKSİK KISIM`)**. İpuçlarını takip ederek bu kısımları doğru Python kodlarıyla tamamlamanız ve modeli başarıyla eğitmeniz beklenmektedir.

Başarılar!

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_theme(style="whitegrid")
```

```python
# Veriyi yükleme ve ön inceleme
df = pd.read_excel('ayakkabi_numarasi.xlsx')

# Cinsiyet sütununu sayısal (1 ve 0) formata çevirme
df['Cinsiyet_E'] = (df['Cinsiyet(E veya K)'] == 'E').astype(int)
df.head()
```

### GÖREV 1: X ve y Değişkenlerinin Ayrılması
Hedef değişkenimiz (bağımlı değişken) `y`, **'Boy(metre)'** sütunudur. Bağımsız değişkenler (`X`) ise **'Kilo(kg)', 'Yaş', 'Cinsiyet_E' ve 'Ayakkabı Numarası'** olmalıdır.

```python
# --- GÖREV 1'i TAMAMLAYIN ---
# X için 'Kilo(kg)', 'Yaş', 'Cinsiyet_E', 'Ayakkabı Numarası' sütunlarını seçin
feature_cols = ['Kilo(kg)', 'Yaş', 'Cinsiyet_E', 'Ayakkabı Numarası']  # Eksik kısımları doldurun
X = df[feature_cols]

# y için 'Boy(metre)' sütununu seçin
y = df['Boy(metre)']  # Eksik kısmı doldurun

print("X boyutu:", X.shape)
print("y boyutu:", y.shape)
```

### GÖREV 2: Veriyi Eğitim ve Test Setlerine Ayırma
Veriyi %80 eğitim, %20 test olacak şekilde ayırın. Bunun için `test_size` parametresini doğru değere ayarlayın.

```python
# --- GÖREV 2'yi TAMAMLAYIN ---
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,  # Buraya %20'yi ifade eden ondalıklı değeri yazın
    random_state=22
)
```

### GÖREV 3: Modelin Kurulması ve Eğitilmesi
Doğrusal Regresyon (LinearRegression) modelini oluşturun ve eğitim verileri ile eğitin.

```python
# --- GÖREV 3'ü TAMAMLAYIN ---
# Modeli çağırın
model = LinearRegression()

# Modeli X_train ve y_train verileri ile eğitin (.fit metodunu kullanın)
model.fit(X_train, y_train)  # Eksik kısımları doldurun

print("Model başarıyla eğitildi!")
```

### GÖREV 4: Tahmin Yapma ve Metriklerin Hesaplanması
Test verisi (`X_test`) üzerinde tahminler üretin ve modelin başarısını $R^2$ skoru ile ölçün.

```python
# --- GÖREV 4'ü TAMAMLAYIN ---
# Test seti üzerinde tahmin yapın (.predict metodunu kullanın)
y_pred = model.predict(X_test)  # Eksik kısmı doldurun

# R² Skorunu hesaplayın
r2 = r2_score(y_test, y_pred)
print(f"Test Seti R² Skoru: {r2:.3f}")
```
