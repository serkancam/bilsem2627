# Python'da `for` Döngüsü, `in` Operatörü ve `range()` Fonksiyonu

Bu çalışma notunda, Python'da tekrarlı işlemleri yönetmek için en sık kullanılan yapılardan biri olan `for` döngüsünü, `in` anahtar kelimesini ve `range()` fonksiyonunu adım adım inceleyeceğiz.

---

## 1. Temel Kavramlar: `for` ve `in` Mantığı

- **`while` döngüsü:** Bir mantıksal koşul doğru (`True`) olduğu sürece çalışır. Sayacı elle artırmamız gerekir.
- **`for` döngüsü:** Belirli bir veri topluluğunun (liste, metin, sayı aralığı) elemanları üzerinde **teker teker gezinmek (iterate etmek)** için kullanılır.
- **`in` anahtar kelimesi:** Topluluğun içindeki her bir elemanı sırayla alıp döngü değişkenine bağlar ("koleksiyon *içindeki* her eleman için").

### 1. Bölüm: Döngüsüz Tekrar ve `for` Döngüsüne İhtiyaç

Eğer döngü kullanmasaydık, 1'den 5'e kadar olan sayıları yazdırmak için tek tek `print()` çağırmak zorunda kalırdık.

```python
# Döngü kullanmadan 1'den 5'e kadar sayılar:
print("Döngüsüz:")
print("Sayı: 1")
print("Sayı: 2")
print("Sayı: 3")
print("Sayı: 4")
print("Sayı: 5")

# Aynı işlemin for döngüsü ile yapılışı:
print("\nFor Döngüsü ile:")
for sayi in range(1, 6):
    print("Sayı:", sayi)
```

---
## 2. `range()` Fonksiyonu ve 3 Farklı Kullanımı

`range()` fonksiyonu sıralı tam sayılar üreten bir yapıdır. Üç temel kullanım biçimi vardır:

1. **`range(bitis)`:** `0`'dan başlar, `bitis` değerine kadar (bitiş **dahil değildir**) 1'er artar.
2. **`range(baslangic, bitis)`:** Belirtilen başlangıçtan başlar, bitiş değerine kadar (bitiş **dahil değildir**) 1'er artar.
3. **`range(baslangic, bitis, adim)`:** Belirtilen başlangıçtan bitişe kadar `adim` değeri kadar artarak ya da azalarak ilerler.

```python
# A) Tek Parametre: range(bitis) -> 0'dan 4'e kadar
print("A) range(5) Kullanımı (0'dan başlar, 5 DAHİL DEĞİL):")
for i in range(5):
    print(i, end=" ")
print()

# B) İki Parametre: range(baslangic, bitis)
print("\nB) range(3, 8) Kullanımı (3'ten başlar, 8 DAHİL DEĞİL):")
for i in range(3, 8):
    print(i, end=" ")
print()

# C) Üç Parametre: range(baslangic, bitis, adim)
print("\nC) range(2, 15, 3) Kullanımı (2'den başlar, 3'er 3'er artar):")
for i in range(2, 15, 3):
    print(i, end=" ")
print()

# D) Geriye Doğru Sayma: Negatif Adım
print("\nD) range(10, 0, -2) Kullanımı (10'dan başlar, 2'şer 2'şer azalır):")
for i in range(10, 0, -2):
    print(i, end=" ")
print()
```

## 3. Metinler (String) ve Listeler Üzerinde Gezinme

`for` döngüsü sadece sayılar için değildir. Karakter dizileri (metinler) ve listeler gibi koleksiyonların tüm elemanları üzerinde indekslerle uğraşmadan doğrudan gezinmeyi sağlar.

```python
# String üzerinde doğrudan gezinme:
mesaj = "BİLSEM"
print(f"'{mesaj}' kelimesinin harfleri:")
for harf in mesaj:
    print("Harf ->", harf)

print("-" * 40)

# Liste üzerinde gezinme ve not ortalaması hesaplama:
notlar = [85, 90, 78, 95, 100]
toplam = 0

print("Öğrenci Notları:", notlar)
for n in notlar:
    toplam = toplam + n
    print("İncelenen not:", n, "| Ara Toplam:", toplam)

ortalama = toplam / len(notlar)
print("Sınıf Not Ortalaması:", ortalama)
```

---
## 4. `enumerate()` ile Hem Sıra (İndeks) Hem Değer Alma

Döngü esnasında hem elemanın kaçıncı sırada olduğunu (indeksini) hem de elemanın kendisini aynı anda kullanmak istediğimizde `enumerate()` fonksiyonu kullanılır.

```python
sehirler = ["Edirne", "İstanbul", "Ankara", "İzmir"]

# enumerate() her adımda (indeks, eleman) ikilisi üretir:
for sira, sehir in enumerate(sehirler, start=1):
    print(f"{sira}. Şehir: {sehir}")
```

## 5. Uygulama Örneği: Çift Sayıların Toplamı

1 ile 20 arasındaki çift sayıları `range(2, 21, 2)` ile adım miktarını kullanarak toplayalım:

```python
cift_toplam = 0

for cift_sayi in range(2, 21, 2):
    cift_toplam = cift_toplam + cift_sayi
    print(f"Eklenen çift sayı: {cift_sayi:2d} | Güncel Toplam: {cift_toplam}")

print(f"\nSonuç: 1 ile 20 arasındaki çift sayıların toplamı = {cift_toplam}")
```
